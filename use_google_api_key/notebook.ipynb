# Gemini API 호출 프로젝트 (google-genai)

이 노트북은 최신 `google-genai` SDK를 사용하여 Gemini 모델을 호출하는 예제입니다.

```python
import os
from dotenv import load_dotenv
from google import genai
```

```python
# 환경변수 로드 및 확인
load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")
model_name = os.getenv("GOOGLE_MODEL", "gemini-2.0-flash-lite")

if not api_key or api_key == "your_api_key_here":
    print("경고: GOOGLE_API_KEY가 설정되지 않았습니다.")
else:
    print(f"API 키 로드 완료 (모델: {model_name})")
```

```python
# prompt.md 파일 읽기
with open("prompt.md", "r", encoding="utf-8") as f:
    prompt_content = f.read()

print("--- 프롬프트 내용 ---")
print(prompt_content)
```

```python
# Gemini API 호출
client = genai.Client(api_key=api_key)

response = client.models.generate_content(
    model=model_name,
    contents=prompt_content
)

print("--- 응답 결과 ---")
print(response.text)
```
