```python
# Create
# Read
# Update
# Delete
# CRUD
# 함수 vs method

# Create & Read
animals = ["고양이", "강아지", "토끼"]
print("원본:", animals)

# Update
animals.append("햄스터")
print("append 후:", animals)

animals.insert(1, "앵무새")
print("insert 후:", animals)

animals[0] = "페르시안 고양이"
print("수정 후:", animals)

# Delete
animals.remove("앵무새")
print("remove 후:", animals)

removed = animals.pop()
print("pop으로 꺼낸 값", removed)
print("pop 후:", animals)

del animals[0]
print("del 후:", animals)
```

```python
scores = [85, 92, 78, 95, 88, 70]
print("원본:", scores)

sorted_scores = sorted(scores)
print("sorted() 결과:", sorted_scores)
print("원본 유지 확인:", scores)  # 원본은 그대로

desc_scores = sorted(scores, reverse = True)
print("내림차순:", desc_scores)
```

```python
scores = {
    "국어": 90,
    "영어": 85,
    "수학": 92,
    "과학": 88
}

for subject, score in scores.items():
    print(f"{subject}: {score}")

for score in scores.values():
    print(f"value: {score}")

for subject in scores.keys():
    print(f"key: {subject}")
```
